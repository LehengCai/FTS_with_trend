```python
#!/usr/bin/env python
# coding: utf-8

# In[50]:


import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import pandas as pd
import os
from matplotlib import cm


# In[51]:


df = pd.read_csv('z1.csv')
df = df.iloc[:, 1:]
z1 = df.values.T
z1.shape

df = pd.read_csv('z2.csv')
df = df.iloc[:, 1:]
z2 = df.values.T

df = pd.read_csv('z3.csv')
df = df.iloc[:, 1:]
z3 = df.values.T

# df = pd.read_csv('z22.csv')
# df = df.iloc[:, 1:]
# z22 = df.values.T

# df = pd.read_csv('z33.csv')
# df = df.iloc[:, 1:]
# z33 = df.values.T

# df = pd.read_csv('z4.csv')
# df = df.iloc[:, 1:]
# z4 = df.values.T


# In[52]:


x = np.linspace(1961, 1994, num=34, dtype=int)
y = np.linspace(1, 365, num=365, dtype=int)
x, y = np.meshgrid(x, y)
x.shape


# In[53]:


fig = plt.figure(figsize=(10,7), facecolor='white')
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(x, y, z3, color='deepskyblue', alpha=0.2)
ax.plot_surface(x, y, z2,  color='deepskyblue', alpha=0.2)

ax.plot_surface(x, y, z1,  color='red', alpha=0.2)

ax.grid(False)



# ax.w_xaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))
# ax.w_yaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))
# ax.w_zaxis.set_pane_color((1.0, 1.0, 1.0, 1.0))


ax.set_xlabel('Year', labelpad=10)
ax.set_ylabel('Day', labelpad=10)
ax.set_zlabel('Temperature', labelpad=10)


ax.set_zlim([-25,25])
ax.view_init(elev=15, azim=-135)
#ax.view_init(elev=15, azim=45)
plt.tight_layout()
plt.savefig('Montreal-temp-95SCR.pdf', format='pdf')
plt.show()




```
